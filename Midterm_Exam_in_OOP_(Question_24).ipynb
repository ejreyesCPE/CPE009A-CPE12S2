```python
#Write a function that checks if a string of parentheses (), brackets [], and braces {} is balanced.

#Examples:

#    "({[]})" → True

#    "([)]" → False

#   "((()))" → True

#    "({)}" → False

def check_parentheses(string):
  for i in string:
    if i == "(" or i == "[" or i == "{":
      string = True
    else:
      return False
    break

    if i == ")" or i == "]" or i == "}":
      string = True
    else:
      return False
  return string

string1 = "({[]})"
print(check_parentheses(string1))
string2 = "([)]"
print(check_parentheses(string2))
string3 = "((()))"
print(check_parentheses(string3))
string4 = "({)}"
print(check_parentheses(string4))
```
