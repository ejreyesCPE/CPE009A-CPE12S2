**LOOP Statements**

FOR LOOP

```python
# Example 1
week = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']

for x in week:
  print(x)
```

```python
# Example 2 (The break statement)
week = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']

for x in week:
  print(x)
  if x == "Thursday":
    break
```

```python
# Example 3 (The break statement)
week = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']

for x in week:
  if x == "Thursday":
    break
  print(x)
```

```python
# Example 4 (Looping through a String)
week = ['Sunday', 'Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday']

for x in "week":
  print(x)
```

The range() Function



*   To loop through a set of code a specified number of times, we can use the range() function.
*   The range() function returns a sequence of numbers, starting from 0 by default, and increments by 1 (by default), and ends at a specified number.





```python
# Example:
for x in range(6):
  print(x)
for x in range(2, 6):
  print(x)
```

```python
for x in range(2, 6, 2):
  print(x)
```

NESTED LOOPS

```python
adjective = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adjective:
  for y in fruits:
    print(x, y)
```

WHILE LOOP

```python
# Example 5 (while loop)
i = 1
while i < 6:
  print(i)
  i += 1 # i = i + 1 Assignment Operator
```

The break Statement

```python
# The break Statement
i = 1
while i < 6:
  print(i)
  if i == 3:
    break # <--
  i += 1
```

The continue Statement

```python
# The continue Statement
i = 0
while i < 6:
  i += 1
  if i == 3:
    continue # <--
  print(i)
```

```python
i = 0
while i < 6:
  i += 1
  if i != 3:
    continue # <--
  print(i)
```

The else Statement

```python
# The else Statement
i = 1
while i < 6:
  print(i)
  i += 1
else:
  print("i is no longer less than 6")
```
