### Dictionaries



```python
# 1. Create a dictionary named contactinfo
contactinfo = {'id':1, 'first_name':'John', 'last_name':'Doe', 'contact_number':'09060611233'}

# 2. Print the following values:
print("contactinfo['id']:", contactinfo['id'])
print("contactinfo['first_name']:", contactinfo['first_name'])
print("contactinfo['last_name']:", contactinfo['last_name'])
print("contactinfo['contact_number']:", contactinfo['contact_number'])

# Attempt to print non-existent key 'age' (this will cause a KeyError)
print("\nAttempting to print contactinfo['age'] (this will cause an error):")
try:
    print("contactinfo['age']:", contactinfo['age'])
except KeyError as e:
    print(f"Error: {e}. Key 'age' does not exist in the dictionary.")

# 3. Type the code: for k,v in contactinfo: print(k) (This iterates over keys only by default)
print("\nIterating over keys (for k,v in contactinfo: print(k)):")
for k in contactinfo:
    print(k) # Original instruction would imply k,v, but direct iteration yields only keys

# 4. Type the code: for k,v in contactinfo.items(): print(k,v)
print("\nIterating over key-value pairs (for k,v in contactinfo.items(): print(k,v)):")
for k,v in contactinfo.items():
    print(k,v)

# 5. Assign new values:
contactinfo['id'] = 2
contactinfo['first_name'] = 'Max'

# 6. Print contactinfo (after assignments)
print("\ncontactinfo after modification:", contactinfo)
```

1. Make a program that would calculate the distance between two points given a list of coordinates. Use the distance formula. coorindates_list = [(1,1), (2,3)]


```python
import math
coordinates_list = [(1,1), (2,3)]

x1 = coordinates_list[0][0]
y1 = coordinates_list[0][1]

x2 = coordinates_list[1][0]
y2 = coordinates_list[1][1]

distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2) # Distance Formula

print(f"The distance between {coordinates_list[0]} and {coordinates_list[1]} is: {distance}")
```

2. For a given string input, replace all the words “stupid” with an asterisk * equal to the length of the string. The new
string value should be displayed with the asterisks.


```python
user_input = input("Input: ")

target = "stupid"

censor = "*" * len(target)

filtered_string = user_input.replace(target, censor)

print("Output:", filtered_string)
```

3. Create a simple phonebook program that can read from a list of dictionaries. The program should be able to display a
person’s name, contact, and address based from a user input which is the id of the record.


```python
phonebook = [ # List
    {"id": "1", "name": "Eli Justin Reyes", "contact": "09494083248", "address": "Blk 3, Lot 4 Bria La Hacienda Teresa, Rizal"},
]

search_id = input("Enter the ID of the record: ") # Finds the record of person by ID

found = False
for record in phonebook:
  if record["id"] == search_id:
        print("-" * 60)
        print(f"ID Number:   {record['id']}")
        print(f"Name:    {record['name'].upper()}")
        print(f"Phone:   {record['contact']}")
        print(f"Address: {record['address']}")
        print("-" * 60)
        found = True
        break

if not found:
    print("Error: Record ID not found.") # Will print this if not registered in phonebook (error)
```
